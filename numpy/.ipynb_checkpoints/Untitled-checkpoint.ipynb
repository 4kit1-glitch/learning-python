```python

# creating a 1d array
nums = [1, 2, 3, 4, 5]
nums_arr = np.array(nums)
print(nums_arr)
```

```python
# lets create a 2d array
row1 = [2, 4, 5]
row2 = [0, 5, 7]
nested_list = [row1, row2]
matrix = np.array(nested_list)
print(matrix)
```

```python
# lets create an array of zero
zeros = np.zeros((3, 7))
ones = np.ones((3, 7))

#lets create an array with linearly spaced values
arr_lsv = np.linspace(0, 100, 25) # (0, 100, 25) means 25 evenly spaced nums from 0 - 100
print(arr_lsv)
print(zeros)
print(ones)
```
