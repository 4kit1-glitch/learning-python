# debugging in oop

1. use the vars to get all attributes
2. use the isinstance to show that an object is an instance of a particular class
3. use the hasatt to show if a particular attribute belongs to a function



```python
class Vehicle:
    """ info on vehicles"""
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age
```

```python
car = Vehicle("car", 2005)

print(isinstance(car, Vehicle)) # check is car is an object of vehicle
print(vars(car))
print(hasattr(car, "name"))
```
