````python
# simple format for the formation of a class

class ClassName:
    """
    Breif description of the class is given here
    Attributes:
        attribute1: description of attribute1
        attribute2: description of attribute2
    methods:
        method1: description of method1
        method2: description of method2
    objects:
        object1: description of object1
        object2: description of object2
    
    note: description of any other important information about the class
    """
    #1. class level variables
    class_variable1 = "value1"
    class_variable2 = "value2"

    def __init__(self, data) -> None:
        """ initialization of class objects"""
        self.data = data

    #3 public methods
    def method1(self):
        """ description of method1"""
        pass

    #4. dunder methods
    def __str__(self):
        """ description of __str__ method"""
        return f"ClassName with data: {self.data}"  
    
    #5. private methods - begins with double underscore
    def __private_method(self):
        """ description of private method"""
        pass
````

````python
# this creates a class object
class Time:
    """Represents the time of the day"""

    # class level variables
    time = 0
    start = 0
    end = 0


````

````python
# creating an output class method

morning = Time() # an object
afternoon = Time() # and object of class time

print(type(morning))
print(morning, afternoon)
````

# OBJECTS ARE MUTIBLE

````python
# OBJECTS are mutible means we can modify the object attributes after they have been created.

class User:
    """represents a user of a system"""
    DEFAULT_ROLE = "guest"
    def __init__(self, name: str, email: str) -> None:
        self.name = name
        self.email = email
        self.role = self.DEFAULT_ROLE

harris = User("harris", "harris@gmail.com")
print(harris.name, harris.email, harris.role)
harris.role = "admin" # we can modify the role attribute of harris object
print(harris.name, harris.email, harris.role)
````

# ENCAPSULATION AND GENERALISATION

````python
def make_admin(user: User) -> None:
    user.role = "admin"
def rename_user(user: User, new_name: str) -> None:
    user.name = new_name

nathalie = User("nathalie", "nathalie@gmail.com")
print(nathalie.name, nathalie.email, nathalie.role)

    
````

````python
# encapsulating and makeing a function that modifies the role of a user object to admin
# showing that objects are mutable and can be modified by functions that take them as arguments 
# also showing generalization of code by creating a function that can modify the role of any user object to admin
make_admin(nathalie)
print(nathalie.name, nathalie.email, nathalie.role)
````

````python
from copy import copy, deepcopy #deep 

# this creates an identical copy but they are different objects in memory
# this is different form deep copy which creates a new object and recursively copies all objects found in the original

kelly = copy(nathalie)
print(kelly.name, kelly.email, kelly.role)
print(nathalie.name, nathalie.email, nathalie.role)

````

the copy function from the copy module copies a method it attributes exactly same as another method but the objects are different in memory

````python
print(nathalie == kelly) # they are equal in value but have different as objects memory addresses
print(nathalie is kelly) # they are not the same object in memory
print(id(nathalie), id(kelly)) # the id function returns the memory address of the object in python, we can see that nathalie and kelly have different memory addresses even though they have the same values for their attributes.
````
